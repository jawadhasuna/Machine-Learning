# **Principal Component Analysis(PCA)**

**Principal Component Analysis (PCA) is a technique used in machine learning and data analysis to reduce the number of features (dimensions) in a dataset while keeping as much important information as possible.**

# **--> PCs**

**Each PC is a mathematical transformation... mathematical function of features.**

**Each Principal Component is a linear combination of original features.**

    PC1 = w1*x1 + w2*x2 + w3*x3 + ... + wn*xn

**Where:**

**x1, x2, x3... = original features**

**w1, w2, w3... = weights (decided by PCA)**

# **What are PC1, PC2, PC3?**

**Principal Components (PCs) are new features created from original features.**

**🔹 PC1 (First Principal Component)**

Captures maximum variance (most important information)

It is the best single summary of data

**🔹 PC2 (Second Principal Component)**

Captures the second highest variance

It is orthogonal (perpendicular) to PC1

**🔹 PC3 (Third Principal Component)**

Captures the third highest variance

Also orthogonal to both PC1 and PC2

*And so on...*

# **How does PCA reduce dimensions?**

**PCA works in these simple steps:**

**1. It looks at the variance (information) in data**

**2. It finds directions where data varies the most**

**3. It creates new axes (called Principal Components)**

**4. It keeps only the most important ones and removes the rest**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer=load_breast_cancer()
```

```python
cancer.keys()
```

```python
print(cancer['DESCR'])
```

```python
df=pd.DataFrame(cancer['data'],columns=cancer['feature_names'])
```

```python
df.head(5)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
scaler.fit(df)
```

```python
scaled_data=scaler.transform(df)
```

```python
scaled_data
```

```python
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2)
```

```python
pca.fit(scaled_data)
```

```python
x_pca=pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
x_pca.shape
```

```python
scaled_data
```

```python
x_pca
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0],x_pca[:,1],c=cancer['target'])
plt.xlabel('First principle component')
plt.ylabel('Second principle component')



```

```python
plt.figure(figsize=(8,6))
plt.scatter(df.iloc[:,0], df.iloc[:,1], c=cancer['target'])
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Original Data (Before PCA)')
plt.show()
```
