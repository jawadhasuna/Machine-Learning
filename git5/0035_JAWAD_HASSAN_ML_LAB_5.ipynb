```python
#libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
```

```python
#activity:
iris=pd.read_csv('/Users/macbook/Desktop/sem6/ML/lab5/git5/Iris.csv')
print(iris.head())
x=iris[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
y=iris['Species']
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
model=SVC()
model.fit(xtrain,ytrain)
predict=model.predict(xtest)
print('Classification Report:')
print(classification_report(ytest,predict))
print('Confusion Matrix:')
print(confusion_matrix(ytest,predict))
```

```python
#task1- 2 features
x=iris[['SepalLengthCm','SepalWidthCm']]
y=iris['Species']
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
model=SVC()
model.fit(xtrain,ytrain)
predict=model.predict(xtest)
print('Classification Report:')
print(classification_report(ytest,predict))
print('Confusion Matrix:')
print(confusion_matrix(ytest,predict))
```

```python
#task2- 4 features
x=iris[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
y=iris['Species']
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
model=SVC() #same as model default = SVC(kernel='rbf', C=1.0, gamma='scale')
model.fit(xtrain,ytrain)
predict=model.predict(xtest)
print('Classification Report:')
print(classification_report(ytest,predict))
print('Confusion Matrix:')
print(confusion_matrix(ytest,predict))
```

```python
#task3- different svm kernels
klist=['linear','poly','rbf']
acc=[]
for k in klist:
    model=SVC(kernel=k)
    model.fit(xtrain,ytrain)
    predict=model.predict(xtest)
    acc.append(accuracy_score(ytest,predict))
#task4- compare results
    print(f'Classification Report for {k} kernel:')
    print(classification_report(ytest,predict))
    print(f'Confusion Matrix for {k} kernel:')
    print(confusion_matrix(ytest,predict))
```

```python
#task5- accuracy of models
plt.bar(klist,acc)
plt.xlabel('SVM Kernel')
plt.ylabel('Accuracy')
plt.title('Accuracy of SVM Kernels on Iris Dataset')
plt.show()
```

## Summary:
### With 2 features the accuracy was down but with 4 features the accuracy of all kernels linear, polynomial and rbf are same because the data set is quite easily linearly separable so not much difference.
