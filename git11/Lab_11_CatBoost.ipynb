## **Introduction**
CatBoost is a Gradient Boosting algorithm developed by Yandex (2017).  
It is designed to handle categorical data efficiently and reduce overfitting.


## **Gradient Boosting vs CatBoost**


- **Gradient Boosting**: Combines weak learners sequentially  and Minimizes error using gradients  
- **CatBoost**: Handles categorical data + reduces overfitting  


## **Mathematical Idea**

Fₘ(x) = Fₘ₋₁(x) + hₘ(x)


Each new model improves previous errors.


## **Handling Categorical Data**


Traditional methods:
- One-hot encoding  
- Label encoding  

CatBoost uses:
- Ordered Target Encoding (no data leakage)
